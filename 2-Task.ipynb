```python
# 2. Write a Python program that takes a user's input for the price of a Zomato order as a string, converts it to a float using type casting,adds 18% GST, and prints the final bill amount.
```

```python
price = int(input("Enter Price off a Zomato order: "))

order_price_casting = float(price)

add_gst = (price * 18)/100

final_price = add_gst + price

print("Actual amount",order_price_casting)
print("Amount of GST is:",add_gst,"%")
print("Final Amount",final_price)
```
