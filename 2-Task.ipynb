```python
# Take the Zomato order price input from the user as a string
price_str = input("Enter the Zomato order price: ")

try:
    # Convert the string input to a float using type casting
    base_amount = float(price_str)
    
    # Define the GST rate (18%)
    gst_rate = 18 
    
    # Calculate the GST amount and final bill amount
    gst_amount = (base_amount * gst_rate) / 100
    final_amount = base_amount + gst_amount
    
    # Display the final bill results
    print(f"\n--- Zomato Order Bill ---")
    print(f"Items Total: ₹{base_amount:.2f}")
    print(f"GST (18%):   ₹{gst_amount:.2f}")
    print(f"Total Pay:   ₹{final_amount:.2f}")

except ValueError:
    # Handle the error if the user enters a non-numerical string
    print("Invalid input! Please enter a valid numerical price.")

```
